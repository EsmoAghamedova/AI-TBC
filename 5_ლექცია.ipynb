ამ ლექციაზე ვისწავლით:
- სიმრავლეები (`set`) და მათი ოპერაციები
- ლექსიკონები (`dict`): key-value სტრუქტურა, მეთოდები, dict comprehension
- ბულის ტიპი (`bool`) და ლოგიკური ოპერატორები (`and`, `or`, `not`)


## 1. სიმრავლეები (set) და მათი ოპერაციები

### რა არის სიმრავლე?

**სიმრავლე (set)** არის მონაცემთა სტრუქტურა, რომელიც ინახავს **უნიკალურ, განუმეორებელ** ელემენტებს, **დაუწესრიგებელი** სახით. ეს ნიშნავს ორ მნიშვნელოვან თავისებურებას:

1. სიმრავლეში **ორჯერ ერთი და იგივე ელემენტი ვერ მოხვდება** — თუ დუბლირებულ მნიშვნელობას შევეცდებით დამატებას, Python ავტომატურად უგულებელყოფს მას.
2. სიმრავლის ელემენტებს **არ აქვთ ინდექსი** — ანუ ვერ მივმართავთ მათ `set[0]`-ის მსგავსი სინტაქსით, რადგან სიმრავლეს არ აქვს ფიქსირებული თანმიმდევრობა.

სიმრავლე იქმნება ფიგურული ფრჩხილებით:

```python
fruits = {"ვაშლი", "ბანანი", "მსხალი"}
```

>  **ყურადღება:** ცარიელი ფიგურული ფრჩხილები `{}` არ ქმნის ცარიელ სიმრავლეს — ის ქმნის ცარიელ ლექსიკონს (dict-ს ვნახავთ შემდეგ ნაწილში). ცარიელი სიმრავლის შესაქმნელად საჭიროა `set()` ფუნქცია.

### სად გამოგვადგება სიმრავლე პრაქტიკაში?

სიმრავლის ყველაზე გავრცელებული გამოყენების შემთხვევებია:
- **დუბლიკატების მოცილება** სიიდან — უბრალოდ სიის სიმრავლედ გადაქცევით.
- **სწრაფი შემოწმება**, არის თუ არა ელემენტი კრებულში (`in` ოპერატორი სიმრავლეზე მუშაობს გაცილებით სწრაფად, ვიდრე დიდ სიაზე).
- **მათემატიკური სიმრავლური ოპერაციები** — გაერთიანება, თანაკვეთა, სხვაობა.

### სიმრავლის ოპერაციები

| ოპერაცია | მეთოდი / სიმბოლო | მაგალითი |
|---|---|---|
| გაერთიანება | `\|` ან `.union()` | `A \| B` |
| თანაკვეთა | `&` ან `.intersection()` | `A & B` |
| სხვაობა | `-` ან `.difference()` | `A - B` |
| სიმეტრიული სხვაობა | `^` ან `.symmetric_difference()` | `A ^ B` |
| ელემენტის დამატება | `.add(x)` | |
| ელემენტის წაშლა | `.remove(x)` ან `.discard(x)` | |


````python
fruits = {"ვაშლი", "ბანანი", "მსხალი", "ვაშლი"}   # დუბლირებული "ვაშლი" ავტომატურად აღმოიფხვრება
print(fruits)
print(len(fruits))
````

````python
# დუბლიკატების მოცილება სიიდან სიმრავლის დახმარებით
numbers_list = [1, 2, 2, 3, 4, 4, 4, 5]
unique_numbers = set(numbers_list)
print(unique_numbers)

# საჭიროების შემთხვევაში, ისევ სიად გადაქცევა
unique_list = list(unique_numbers)
print(unique_list)
````

````python
# ელემენტების დამატება და წაშლა
colors = {"წითელი", "მწვანე"}

colors.add("ლურჯი")
print(colors)

colors.remove("მწვანე")
print(colors)
````

````python
# სიმრავლური ოპერაციები
students_math = {"ნინო", "გიორგი", "ანა", "დათო"}
students_physics = {"გიორგი", "ლუკა", "ანა"}

print("ორივე საგანზე დარეგისტრირებული:", students_math & students_physics)   # თანაკვეთა
print("ერთ-ერთ საგანზე მაინც:", students_math | students_physics)              # გაერთიანება
print("მხოლოდ მათემატიკაზე:", students_math - students_physics)               # სხვაობა
````

````python
# სწრაფი შემოწმება - არის თუ არა ელემენტი სიმრავლეში
allowed_users = {"admin", "moderator", "editor"}

print("admin" in allowed_users)
print("guest" in allowed_users)
````

## 2. ლექსიკონები (dict): key-value სტრუქტურა, მეთოდები, dict comprehension

### რა არის ლექსიკონი?

**ლექსიკონი (dictionary, dict)** არის მონაცემთა სტრუქტურა, რომელიც ინახავს ინფორმაციას **წყვილების** სახით — **გასაღები (key)** და **მნიშვნელობა (value)**. ეს კონცეფცია ძალიან ჰგავს რეალურ ლექსიკონს: სიტყვას (გასაღებს) შეესაბამება მისი განმარტება (მნიშვნელობა).

ლექსიკონი იქმნება ფიგურული ფრჩხილებით, სადაც თითოეული წყვილი იწერება `გასაღები: მნიშვნელობა` ფორმატით:

```python
student = {
    "name": "ნინო",
    "age": 16,
    "grade": "10ბ"
}
```

ინფორმაციის მისაწვდომად, სიისგან განსხვავებით (სადაც რიცხვით ინდექსს ვიყენებდით), აქ ვიყენებთ **გასაღებს**:

```python
print(student["name"])   # ნინო
```

**მნიშვნელოვანი წესები:**
- გასაღები ლექსიკონში უნდა იყოს **უნიკალური** — თუ ერთსა და იმავე გასაღებს ხელახლა მივანიჭებთ მნიშვნელობას, ძველი მნიშვნელობა უბრალოდ გადაიწერება.
- გასაღები უნდა იყოს **უცვლელი (immutable)** ტიპის — შეიძლება იყოს სტრიქონი, რიცხვი ან ტუპლი, მაგრამ **არა** სია.
- მნიშვნელობა შეიძლება იყოს **ნებისმიერი** ტიპის — მათ შორის სია, ან თუნდაც სხვა ლექსიკონი.

### ხშირად გამოყენებული ლექსიკონის მეთოდები

| მეთოდი | დანიშნულება |
|---|---|
| `.keys()` | აბრუნებს ყველა გასაღებს |
| `.values()` | აბრუნებს ყველა მნიშვნელობას |
| `.items()` | აბრუნებს ყველა (გასაღები, მნიშვნელობა) წყვილს |
| `.get(key, default)` | აბრუნებს მნიშვნელობას გასაღებით — შეცდომის ნაცვლად, თუ არ არსებობს, აბრუნებს `default`-ს |
| `.pop(key)` | შლის და აბრუნებს მნიშვნელობას მითითებული გასაღებით |
| `.update({...})` | ამატებს ან განაახლებს რამდენიმე წყვილს ერთდროულად |

### dict comprehension

ისევე, როგორც სიის შემთხვევაში ვისწავლეთ list comprehension, ლექსიკონისთვისაც არსებობს ანალოგიური, კომპაქტური სინტაქსი:

```python
{გასაღები: მნიშვნელობა for ელემენტი in მიმდევრობა}
```


````python
student = {
    "name": "ნინო",
    "age": 16,
    "grade": "10ბ"
}

print(student)
print(student["name"])
print(student["age"])
````

````python
# ახალი წყვილის დამატება და არსებულის განახლება
student = {"name": "ნინო", "age": 16}

student["grade"] = "10ბ"     # ახალი გასაღები-მნიშვნელობის დამატება
print(student)

student["age"] = 17          # არსებული მნიშვნელობის განახლება
print(student)
````

````python
# .get() მეთოდის უსაფრთხო გამოყენება
student = {"name": "ნინო", "age": 16}

print(student.get("grade", "მიუწერელია"))   # გასაღები არ არსებობს - ბრუნდება default მნიშვნელობა
print(student.get("name", "უცნობია"))       # გასაღები არსებობს - ბრუნდება ნამდვილი მნიშვნელობა

# შედარებისთვის: student["grade"] პირდაპირ გამოძახება KeyError-ს გამოიწვევდა
````

````python
# .keys(), .values(), .items() მეთოდები
student = {"name": "ნინო", "age": 16, "grade": "10ბ"}

print(list(student.keys()))
print(list(student.values()))
print(list(student.items()))

for key, value in student.items():
    print(f"{key}: {value}")
````

````python
# ლექსიკონი, რომლის მნიშვნელობაც სია ან სხვა ლექსიკონია
school_data = {
    "student": "გიორგი",
    "grades": [88, 92, 79],
    "contact": {"phone": "555-11-22-33", "city": "თბილისი"}
}

print(school_data["grades"])
print(school_data["contact"]["city"])   # ჩადგმული ლექსიკონის მისაწვდომობა
````

````python
# dict comprehension - რიცხვისა და მისი კვადრატის შესაბამისობა
squares_dict = {x: x ** 2 for x in range(1, 6)}
print(squares_dict)

# dict comprehension სიტყვებისა და მათი სიგრძის შესაბამისობით
words = ["python", "list", "dict", "set"]
word_lengths = {word: len(word) for word in words}
print(word_lengths)
````

## 3. ბულის ტიპი (bool) და ლოგიკური ოპერატორები (and, or, not)

### bool ტიპის გახსენება

**ბულის ტიპი (bool)** არის ერთ-ერთი ყველაზე მარტივი, მაგრამ ფუნდამენტური მონაცემთა ტიპი — მას მხოლოდ **ორი** შესაძლო მნიშვნელობა აქვს: `True` (ჭეშმარიტი) და `False` (მცდარი). ეს მნიშვნელობები ყველაზე ხშირად მიიღება შედარების ოპერატორების გამოყენებისას (`==`, `!=`, `>`, `<` და ა.შ.), რომლებიც უკვე შევისწავლეთ.

```python
is_adult = 18 >= 16
print(is_adult)   # True
```

### bool ტიპის კავშირი მონაცემთა ახლადნასწავლ სტრუქტურებთან

საინტერესოა, რომ `bool` ტექნიკურად წარმოადგენს `int`-ის ერთგვარ განშტოებას — `True` ტოლფასია `1`-ის, ხოლო `False` — `0`-ის:

```python
print(True + True)   # 2
```

ასევე მნიშვნელოვანია ვიცოდეთ, რომ Python-ში ბევრი მნიშვნელობა ავტომატურად აღიქმება, როგორც "ცრუ" (falsy) ლოგიკურ კონტექსტში, თუნდაც ის ფორმალურად `bool` ტიპის არ იყოს:

- ცარიელი სტრიქონი `""`, ცარიელი სია `[]`, ცარიელი ლექსიკონი `{}`, ცარიელი სიმრავლე `set()`
- რიცხვი `0` და `0.0`
- სპეციალური მნიშვნელობა `None`

ყველა დანარჩენი მნიშვნელობა ("არაცარიელი") ითვლება ჭეშმარიტად (truthy).

### ლოგიკური ოპერატორები ახალ სტრუქტურებთან ერთად

ლოგიკური ოპერატორები `and`, `or`, `not`, რომლებიც უკვე გავეცანით მე-2 ლექციაზე, განსაკუთრებით სასარგებლო ხდება ახლა, როცა ვმუშაობთ სიმრავლეებთან და ლექსიკონებთან — მაგალითად, რამდენიმე პირობის ერთდროულად შესამოწმებლად.


````python
# bool მნიშვნელობები შედარების შედეგად
age = 17
has_permission = True

is_adult = age >= 18
print(is_adult, type(is_adult))

can_enter = is_adult and has_permission
print(can_enter)
````

````python
# bool და int-ის კავშირი
print(True + True)     # 2
print(False + 5)       # 5
print(int(True))       # 1
print(bool(0))         # False
print(bool(15))        # True
````

````python
# "ცრუ" (falsy) მნიშვნელობების დემონსტრირება
print(bool(""))         # False - ცარიელი სტრიქონი
print(bool([]))         # False - ცარიელი სია
print(bool({}))         # False - ცარიელი ლექსიკონი
print(bool(0))          # False

print(bool("გამარჯობა"))   # True - არაცარიელი სტრიქონი
print(bool([1, 2]))         # True - არაცარიელი სია
````

````python
# ლოგიკური ოპერატორები ლექსიკონთან ერთად
student = {"name": "ანა", "age": 17, "has_id": True}

can_register = student["age"] >= 16 and student["has_id"]
print(can_register)

# ლოგიკური ოპერატორი სიმრავლის შემოწმებასთან ერთად
allowed_subjects = {"მათემატიკა", "ფიზიკა", "ინფორმატიკა"}
chosen_subject = "ქიმია"

is_allowed = chosen_subject in allowed_subjects
print(f"საგანი დაშვებულია? {is_allowed}")

if not is_allowed:
    print("საგანი არ არის დაშვებული")
````

## საკლასო სავარჯიშოები (დამოუკიდებელი მუშაობისთვის)

თითოეული სავარჯიშოსთვის გამოყენებულია ცალკე უჯრედი — დაწერეთ თქვენი კოდი შესაბამის ადგილას.

---

**სავარჯიშო 1. სიმრავლის ოპერაციები**

მოცემულია ორი სია მოსწავლეების სახელებით, რომლებმაც აირჩიეს ორი სხვადასხვა წრე:

```python
chess_club = ["ნინო", "დათო", "გიორგი", "ლუკა"]
robotics_club = ["გიორგი", "ანა", "დათო"]
```

გადაიყვანეთ ისინი სიმრავლეებად და გამოთვალეთ: 1) რომელი მოსწავლეები დადიან ორივე წრეზე, 2) ვინ დადის მხოლოდ ჭადრაკის წრეზე.


````python
chess_club = ["ნინო", "დათო", "გიორგი", "ლუკა"]
robotics_club = ["გიორგი", "ანა", "დათო"]

chess_set = set(chess_club)
robotics_set = set(robotics_club)

both_clubs = chess_set & robotics_set
only_chess = chess_set - robotics_set

print(both_clubs)
print(only_chess)
````

**სავარჯიშო 2. ლექსიკონის შექმნა და გამოყენება**

შექმენით ლექსიკონი `book`, რომელიც შეიცავს წიგნის სათაურს, ავტორს და გამოცემის წელს. დაამატეთ ახალი გასაღები `"pages"` (გვერდების რაოდენობა) და გამოიტანეთ მთელი ლექსიკონი `.items()`-ის გამოყენებით, თითოეული წყვილი ცალკე ხაზზე.


````python
book = {
    "title": "ვეფხისტყაოსანი",
    "author": "შოთა რუსთაველი",
    "year": 1200
}

book["pages"] = 400

for key, value in book.items():
    print(f"{key}: {value}")
````

**სავარჯიშო 6 (დამატებითი, გაძლიერებული სირთულის).**

მოცემულია მოსწავლეთა ინფორმაცია სიის სახით, სადაც თითოეული მოსწავლე წარმოდგენილია ლექსიკონად:

```python
students = [
    {"name": "ნინო", "grade": 92},
    {"name": "გიორგი", "grade": 78},
    {"name": "ანა", "grade": 85}
]
```

დაწერეთ პროგრამა, რომელიც:
1. list comprehension-ის გამოყენებით ქმნის სიმრავლეს მხოლოდ იმ მოსწავლეების სახელებით, ვისი ქულაც 80-ზე მეტია.
2. dict comprehension-ის გამოყენებით ქმნის ახალ ლექსიკონს, სადაც გასაღებია მოსწავლის სახელი, ხოლო მნიშვნელობა — ბულის ტიპის პასუხი კითხვაზე "ჩააბარა თუ არა" (ქულა >= 80).


````python
students = [
    {"name": "ნინო", "grade": 92},
    {"name": "გიორგი", "grade": 78},
    {"name": "ანა", "grade": 85}
]

high_achievers = {s["name"] for s in students if s["grade"] > 80}
print(high_achievers)

pass_fail_dict = {s["name"]: (s["grade"] >= 80) for s in students}
print(pass_fail_dict)
````
